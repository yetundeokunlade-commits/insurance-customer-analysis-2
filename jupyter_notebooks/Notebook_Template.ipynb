## 1.12 Statistics Fundamentals

Statistics helps me understand, summarise and make decisions from data.

* **Mean:** The mean is the average value. It is calculated by adding all values and dividing by the number of values. In this project, descriptive statistics helped me understand the typical values of the variables in the dataset.

* **Median:** The median is the middle value when data is arranged in order. It can be useful when data contains unusually high or low values because it is less affected by extreme values than the mean.

* **Standard deviation:** Standard deviation measures how spread out values are around the mean. A small standard deviation means values are closer to the mean, while a larger standard deviation means the values are more spread out.

* **Variance:** Variance measures how far data values are spread out from the mean by calculating the average of their squared differences from it. It is closely related to standard deviation, which expresses spread in the original units of measurement.


* **Hypothesis testing:** Hypothesis testing is used to assess whether there is enough statistical evidence to support a relationship or difference observed in data. In this project, I used Chi-square tests of independence to investigate whether selected customer characteristics were associated with caravan insurance ownership.

* **Probability:** Probability describes how likely an event is to occur. It ranges from 0 to 1, or from 0% to 100%. In this project, the distribution of the `CARAVAN` target variable showed that 5.98% of records had caravan insurance, while 94.02% did not.

These statistical principles provide a foundation for understanding the dataset, identifying patterns and evaluating whether observed relationships provide sufficient statistical evidence.

## Section 3: Statistical Analysis

In this section, I use statistical tests to check whether the patterns I found during the exploratory data analysis show evidence of a link with caravan insurance ownership.

The earlier analysis showed differences across age groups, purchasing power groups, car ownership groups, household size groups, and higher education percentage groups.

However, seeing a difference in the data does not automatically mean that there is a real association. The differences could have happened by chance.

I will use statistical tests to check whether there is enough evidence of an association between these selected variables and the `CARAVAN` target variable.

The results will help me understand which patterns have statistical evidence to support them and which do not.


### 3.1 Hypothesis: Age Group and Caravan Insurance

From my earlier analysis, I noticed some differences in caravan insurance ownership across the age groups. I want to check whether there is enough statistical evidence to support a link between age group and caravan insurance ownership.

I will compare `MGEMLEEF` (age group) with `CARAVAN` (caravan insurance ownership).

**Null hypothesis (H₀):**

There is no association between age group and caravan insurance ownership.

**Alternative hypothesis (H₁):**

There is an association between age group and caravan insurance ownership.

I will use a **Chi-square test of independence** to test these hypotheses, with a **5% significance level (0.05)**.

* If the p-value is **less than 0.05**, I will reject the null hypothesis because there is enough statistical evidence of an association.
* If the p-value is **0.05 or higher**, I will not reject the null hypothesis because there is not enough statistical evidence of an association.


### 3.2 Chi-square Test: Age Group and Caravan Insurance

I will use the **Chi-square test of independence** to check whether there is a statistically significant association between age group and caravan insurance ownership.

The test uses the number of records in each age group, including whether they have caravan insurance (`CARAVAN = 1`) or do not have it (`CARAVAN = 0`).

### What this code is doing

1. `from scipy.stats import chi2_contingency` → imports the Chi-square test from SciPy.
2. `age_caravan_table.iloc[:-1, :-1]` → selects the age group and caravan insurance counts, leaving out the total row and total column.
3. `chi2_contingency()` → carries out the Chi-square test.
4. `chi2_stat` → stores the Chi-square test statistic.
5. `p_value` → stores the p-value, which helps me decide whether there is enough evidence of an association.
6. `dof` → stores the degrees of freedom used in the test.
7. `expected` → stores the counts expected if there were no association between the two variables.
8. `print()` → displays the test results.


```python
# Import the Chi-square test of independence from SciPy.
from scipy.stats import chi2_contingency

# Remove the total row and total column from the earlier cross-tabulation.
age_table = age_caravan_table.iloc[:-1, :-1]

# Perform the Chi-square test and store the test results.
chi2_stat, p_value, dof, expected = chi2_contingency(age_table)

# Display the Chi-square test statistic.
print("Chi-square statistic:", round(chi2_stat, 4))

# Display the p-value used to assess the hypothesis.
print("p-value:", round(p_value, 4))

# Display the degrees of freedom for the test.
print("Degrees of freedom:", dof)
```

### Interpretation

The Chi-square test produced a **Chi-square statistic of 3.2919** and a **p-value of 0.6551**.

I used a significance level of **0.05 (5%)**. Since the p-value of **0.6551 is greater than 0.05**, I do not reject the null hypothesis.

The null hypothesis stated that there is no association between age group and caravan insurance ownership.

This means that there is **not enough statistical evidence to conclude that age group is associated with caravan insurance ownership** in this dataset.

Although the earlier exploratory analysis showed some differences in caravan insurance percentages across the age groups, the Chi-square test found that these differences were **not statistically significant** at the 5% level. The observed differences could be due to chance.

This result shows why statistical testing is important. Charts help me see differences between groups, while a statistical test helps me assess whether there is enough evidence of an association.

The next test will examine another variable identified during the exploratory analysis.

The ROC-AUC score ranges from 0 to 1:

* **A score of 0.5** indicates discrimination similar to random ranking.
* **A score closer to 1.0** indicates a stronger ability to distinguish between the two classes.
* **A score below 0.5** indicates discrimination worse than random ranking, assuming the positive class and score direction are correctly defined.

ROC-AUC is useful for this dataset because it measures the model's ability to rank positive records above negative records across different thresholds. It provides information beyond accuracy, although it does not by itself show whether the model's predictions are precise enough for practical customer targeting.

I will calculate the ROC-AUC score using the predicted probabilities from the Logistic Regression model for the testing data.


```python
# Import the ROC-AUC score from Scikit-learn.
from sklearn.metrics import roc_auc_score

# Generate predicted probabilities for the positive class (CARAVAN = 1).
y_prob = logistic_model.predict_proba(X_test_scaled)[:, 1]

# Calculate the ROC-AUC score using the actual outcomes and predicted probabilities.
roc_auc = roc_auc_score(y_test, y_prob)

# Display the ROC-AUC score rounded to four decimal places.
print("ROC-AUC score:", round(roc_auc, 4))
```

### Interpretation

The Logistic Regression model achieved a **ROC-AUC score of 0.7424**.

A ROC-AUC score of **0.5** indicates discrimination similar to random ranking, while a score closer to **1.0** indicates a stronger ability to distinguish between the two classes.

The score of **0.7424** indicates that the model has a **moderate ability to distinguish between records with and without caravan insurance**. This suggests that the model has learned useful patterns from the available data and performs better than random ranking.

However, the ROC-AUC score should be considered alongside the other model evaluation results.

The model achieved **67% recall** for records with caravan insurance, meaning it correctly identified approximately 67% of the records that actually had caravan insurance in the test set.

However, its **precision for this class was only 13%**, meaning that only a small proportion of the records predicted to have caravan insurance actually belonged to that class. Many of the model's positive predictions were false positives.

Therefore, although the model shows some ability to distinguish between the two classes, its relatively low precision means it would need further improvement before being considered reliable for practical customer targeting.

Overall, the Logistic Regression model provides a useful starting point for predicting caravan insurance ownership. However, its limitations must be considered before using its predictions to guide business decisions.
