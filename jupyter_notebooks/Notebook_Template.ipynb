### 1.9 Target Variable Percentage Distribution

The target variable contains two categories: `0` and `1`. I calculate the percentage of records in each category to understand the size of each group more clearly.

This is useful because the target variable is not evenly distributed. Understanding this imbalance is important before selecting and evaluating a machine learning model.

**What this code is doing:**

1. `value_counts()` → counts the number of records in each target category.

2. `normalize=True` → converts the counts into proportions of the total records.

3. `.mul(100)` → converts the proportions into percentages.

4. `.sort_index()` → displays the categories in numerical order.

5. `.round(2)` → rounds the percentages to two decimal places.


```python
# Calculate the percentage of records in each CARAVAN target category.
caravan_percentage = (
    df["CARAVAN"]
    .value_counts(normalize=True)
    .mul(100)
    .sort_index()
    .round(2)
)

print("CARAVAN target percentage distribution:")

print(caravan_percentage)
```

Only **5.98%** of the records have a mobile home/caravan insurance policy (`CARAVAN = 1`).

This shows that the target variable is **not evenly balanced**. Most records are in category `0`, while only a small proportion are in category `1`.

This is important for the machine learning part of the project. A model could achieve high accuracy by predicting `0` for most records because `0` is the largest group.

Therefore, I should not use accuracy alone to judge how well the model performs. I will also consider measures such as **precision, recall, F1-score and the confusion matrix**.

This result agrees with the earlier count of **5,474 records without caravan insurance and 348 records with caravan insurance**.

I will take this imbalance into account when selecting and evaluating the machine learning model.

### 2.3 Caravan Insurance Ownership by Age Group

I compare the `MGEMLEEF` age groups with the `CARAVAN` target variable.

This will show the number of records in each age group who do and do not have caravan insurance.

The result will help me see whether caravan insurance ownership appears to differ between age groups.

**What this code is doing:**

1. `pd.crosstab()` → creates a table showing the relationship between two variables.

2. `df["MGEMLEEF"]` → provides the customer age-group codes.

3. `df["CARAVAN"]` → provides the caravan insurance outcome.

4. `margins=True` → adds row and column totals to the table.

5. `print()` → displays the results.


```python
# Create a cross-tabulation showing caravan insurance ownership by age group.
age_caravan_table = pd.crosstab(
    df["MGEMLEEF"],
    df["CARAVAN"],
    margins=True
)

print("Caravan insurance ownership by age group:")

print(age_caravan_table)
```

### 2.4 Caravan Insurance Percentage by Age Group

The previous table showed the number of records with and without caravan insurance in each age group.

Because the age groups contain different numbers of records, I calculated the percentage of records with caravan insurance within each age group.

This gives a fairer comparison between the age groups because it considers the size of each group.

**What this code is doing:**

1. `pd.crosstab()` → creates a table comparing age group with caravan insurance ownership.

2. `normalize="index"` → calculates the proportion within each age group.

3. `.mul(100)` → converts the proportions into percentages.

4. `.round(2)` → rounds the results to two decimal places.


---

```python
# Calculate the percentage of caravan insurance outcomes within each age group.
age_caravan_percentage = (
    pd.crosstab(
        df["MGEMLEEF"],
        df["CARAVAN"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

print("Caravan insurance percentage by age group:")

print(age_caravan_percentage)
```

### Interpretation

The results show the percentage of records with and without caravan insurance within each age group.

The percentage of records with caravan insurance (`CARAVAN = 1`) is:

* **Age group 1:** 1.35%
* **Age group 2:** 5.99%
* **Age group 3:** 6.10%
* **Age group 4:** 5.96%
* **Age group 5:** 6.22%
* **Age group 6:** 3.33%

Age group 5 has the highest percentage of records with caravan insurance at **6.22%**, followed closely by age group 3 at **6.10%**. Age groups 2 and 4 also have similar percentages at **5.99%** and **5.96%** respectively.

Age group 1 has the lowest percentage at **1.35%**, while age group 6 has **3.33%**.

Overall, caravan insurance ownership appears to differ between the age groups. The largest difference is between age group 1 and age group 5, while age groups 2 to 5 have relatively similar percentages.

However, these results are based on descriptive analysis only. They show patterns in the dataset but do not tell me whether the differences are statistically significant.

A statistical test will be used later in the project to determine whether there is evidence of a relationship between age group and caravan insurance ownership.

This analysis provides an initial answer to the business question and helps identify age group as a variable that should be considered further in the statistical analysis and machine learning stages.

## Section 3: Statistical Analysis

In this section, I use statistical methods to investigate whether the patterns identified during the exploratory data analysis provide evidence of an association between selected customer characteristics and caravan insurance ownership.

The exploratory analysis showed differences in caravan insurance ownership across age groups, purchasing power groups, car ownership groups, household size groups, and higher education percentage groups.

However, descriptive analysis alone cannot tell us whether these differences are statistically meaningful or could have occurred by chance.

I will therefore use appropriate statistical tests to investigate the associations between selected variables and the `CARAVAN` target variable.

The results will help support or challenge the findings from the exploratory data analysis.


### 3.1 Hypothesis: Age Group and Caravan Insurance

Based on the exploratory analysis, I want to test whether there is an association between customer age group and caravan insurance ownership.

The statistical test will compare the `MGEMLEEF` age groups with the `CARAVAN` target variable.

**Null hypothesis (H₀):**

There is no association between age group and caravan insurance ownership.

**Alternative hypothesis (H₁):**

There is an association between age group and caravan insurance ownership.

I will use a **Chi-square test of independence** to test these hypotheses.

The significance level will be set at **5% (0.05)**.

If the p-value is less than 0.05, I will reject the null hypothesis and conclude that there is evidence of an association between age group and caravan insurance ownership.

If the p-value is greater than or equal to 0.05, I will not reject the null hypothesis because there is not enough evidence to conclude that an association exists.


### 3.2 Chi-square Test: Age Group and Caravan Insurance

I will use the Chi-square test of independence to determine whether there is evidence of an association between age group and caravan insurance ownership.

The test will use the observed counts of records in each age group and their caravan insurance outcome.

**What this code is doing:**

1. `from scipy.stats import chi2_contingency` → imports the Chi-square test from SciPy.

2. `age_caravan_table.iloc[:-1, :-1]` → selects the age group and caravan insurance counts from the earlier table, excluding the total row and total column.

3. `chi2_contingency()` → performs the Chi-square test of independence.

4. `chi2_stat` → stores the Chi-square test statistic.

5. `p_value` → stores the p-value used to make the decision about the hypothesis.

6. `dof` → stores the degrees of freedom for the test.

7. `expected` → stores the expected counts if there were no association between the two variables.

8. `print()` → displays the test results.


```python
# Import the Chi-square test of independence from SciPy.
from scipy.stats import chi2_contingency

# Remove the total row and total column from the earlier cross-tabulation.
age_table = age_caravan_table.iloc[:-1, :-1]

# Perform the Chi-square test and store the test results.
chi2_stat, p_value, dof, expected = chi2_contingency(age_table)

# Display the Chi-square test statistic.
print("Chi-square statistic:", round(chi2_stat, 4))

# Display the p-value used to assess the hypothesis.
print("p-value:", round(p_value, 4))

# Display the degrees of freedom for the test.
print("Degrees of freedom:", dof)
```

### Interpretation

The Chi-square test produced a **Chi-square statistic of 3.2919** and a **p-value of 0.6551**.

I used a significance level of **0.05 (5%)**.

The p-value of **0.6551 is greater than 0.05**. Therefore, I **do not reject the null hypothesis**.

The null hypothesis stated that:

> There is no association between age group and caravan insurance ownership.

Based on the test result, there is **not enough statistical evidence to reject the null hypothesis**.

In simple terms, the test does not provide enough evidence to conclude that age group is associated with caravan insurance ownership in this dataset.

Although the earlier exploratory analysis showed some differences in caravan insurance percentages between the age groups, the Chi-square test suggests that these differences are **not statistically significant** at the 5% significance level. This means that the differences observed in the sample could have occurred by chance.

Therefore, the statistical analysis does **not provide sufficient evidence of an association between age group and caravan insurance ownership** in this dataset.

This shows why statistical testing is important. Charts can show differences between groups, but a statistical test helps determine whether there is enough evidence to consider those differences statistically meaningful.

## Section 4: Machine Learning

In this section, I will use machine learning to investigate whether customer characteristics can be used to predict caravan insurance ownership.

The target variable is `CARAVAN`, which has two possible outcomes:

* `0` → customer does not have caravan insurance
* `1` → customer has caravan insurance

Because the target has two possible outcomes, this is a **binary classification problem**.

The machine learning model will learn patterns from the available customer data and use these patterns to predict caravan insurance ownership.

The statistical analysis identified purchasing power, car ownership percentage group and higher education percentage group as variables with statistically significant associations with caravan insurance ownership.

However, machine learning can consider multiple variables together rather than examining each variable separately.

The model's performance will be evaluated using appropriate classification measures. Because the target variable is imbalanced, accuracy alone will not be used to judge the model.

The results will help me assess whether the machine learning approach can provide useful predictions of caravan insurance ownership in this dataset.


### 4.2 Preparing the Data

Before training the machine learning model, I need to separate the target variable from the predictor variables.

The target variable is `CARAVAN`, which is what the model will try to predict.

The remaining variables will be used as predictor variables.

I will create:

* `X` → predictor variables used by the model.
* `y` → target variable containing caravan insurance ownership.

I will then check the shape of both datasets to confirm that the data has been separated correctly.

```python
# Create X by removing the target variable from the dataset.
X = df.drop(columns=["CARAVAN"])

# Create y using the CARAVAN column as the target variable.
y = df["CARAVAN"]

# Check the shape of the predictor variables.
print("Predictor variables (X):", X.shape)

# Check the shape of the target variable.
print("Target variable (y):", y.shape)
```

### 4.3 Train-Test Split

I will split the dataset into training and testing sets before building the machine learning model.

I will use **80% of the data for training** and **20% for testing**.

The training data will be used to teach the model patterns in the customer data.

The testing data will be kept separate and used to evaluate how well the model performs on records it has not seen during training.

I will use `stratify=y` because the `CARAVAN` target variable is imbalanced. This helps maintain a similar proportion of records with and without caravan insurance in both the training and testing sets.

I will also use `random_state=42` so that the same split can be reproduced when the notebook is run again.


```python
# Import the function used to split the data into training and testing sets.
from sklearn.model_selection import train_test_split

# Split the predictor and target data into training and testing sets.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Check the shape of the training predictor data.
print("Training predictor data:", X_train.shape)

# Check the shape of the testing predictor data.
print("Testing predictor data:", X_test.shape)

# Check the shape of the training target data.
print("Training target data:", y_train.shape)

# Check the shape of the testing target data.
print("Testing target data:", y_test.shape)
```

### 4.4 Feature Scaling

The predictor variables have different numerical ranges.

For example, some variables have relatively small values, while others have larger values.

Logistic Regression can perform better when the predictor variables are placed on a similar scale.

I will therefore use `StandardScaler` from Scikit-learn to standardise the predictor variables.

The scaler will be fitted using the training data only. The scaling information learned from the training data will then be applied to both the training and testing data.

This helps prevent information from the testing data being used when training the model and helps avoid data leakage.


```python
# Import StandardScaler to standardise the predictor variables.
from sklearn.preprocessing import StandardScaler

# Create the scaler.
scaler = StandardScaler()

# Fit the scaler using the training data and standardise the training data.
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling learned from the training data to the testing data.
X_test_scaled = scaler.transform(X_test)

# Check the shape of the scaled training data.
print("Scaled training data:", X_train_scaled.shape)

# Check the shape of the scaled testing data.
print("Scaled testing data:", X_test_scaled.shape)
```

### 4.5 Training the Logistic Regression Model

The training data is now ready for the machine learning model.

I will use Logistic Regression from Scikit-learn to learn patterns between the predictor variables and caravan insurance ownership.

The model will be trained using the scaled training data and the known `CARAVAN` outcomes.

The `class_weight="balanced"` setting is used because the target variable is imbalanced. There are many more records without caravan insurance than records with caravan insurance.

This setting gives greater weight to the minority class during training, which may help the model identify records with caravan insurance.

After training, I will use the model to make predictions on the testing data and then evaluate its performance.


```python
# Import Logistic Regression from Scikit-learn.
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model with balanced class weights.
logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

# Train the model using the scaled training data and known target outcomes.
logistic_model.fit(X_train_scaled, y_train)

# Confirm that the model has been trained successfully.
print("Logistic Regression model trained successfully.")
```

### 4.6 Making Predictions

The Logistic Regression model has now been trained using the training data.

I will use the trained model to predict caravan insurance ownership for the testing data.

The model will produce a predicted outcome of either:

* `0` → predicted not to have caravan insurance
* `1` → predicted to have caravan insurance

The predictions will then be compared with the actual `CARAVAN` values in the testing data when evaluating the model.


```python
# Use the trained model to predict caravan insurance ownership for the testing data.
y_pred = logistic_model.predict(X_test_scaled)

# Check the number of predictions generated.
print("Number of predictions:", len(y_pred))

# Confirm that the prediction step completed successfully.
print("Predictions made successfully.")
```

The classification report provides several measures of model performance:

* **F1-score** provides a balance between precision and recall.
* **Support** shows the number of actual records in each class.

These measures are particularly important for this dataset because caravan insurance ownership is much less common than non-ownership.

Therefore, accuracy alone would not give a complete picture of how well the model identifies records with caravan insurance.


```python
# Import the classification report from Scikit-learn.
from sklearn.metrics import classification_report

# Compare the actual test outcomes with the model's predictions.
print(classification_report(y_test, y_pred))
```
